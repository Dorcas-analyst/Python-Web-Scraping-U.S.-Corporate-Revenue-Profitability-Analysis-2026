```python
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://grokipedia.com/page/List_of_largest_companies_in_the_United_States_by_revenue"

page = requests.get(url)

soup = BeautifulSoup(page.text, 'html')
```

```python
print(soup.prettify())
```

```python
soup.find_all('table')[1]
```

```python
table2 = soup.find_all('table')[1]
```

```python
table2.find_all('th')
```

```python
table_header2 = table2.find_all('th')
```

```python
world_table_header2 = [title.text for title in table_header2]
```

```python
world_table_header2
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = world_table_header2)
```

```python
df
```

```python
column_data = table2.find_all('tr')
```

```python
for row in column_data[1:]:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip() for data in row_data]

    length = len(df)
    df.loc[length] = individual_row_data
```

```python
"""
Task 78 - Web Scraping Project
Exact chart-generation code used in the technical report.

Requires: matplotlib  (pip install matplotlib)
Run this after your scraping + DataFrame-building code, or standalone
using the hardcoded data below (taken directly from your scraped tables).
"""

import matplotlib.pyplot as plt

# =========================================================
# CHART 1 — Top 10 U.S. Companies by Revenue (from Table 1)
# =========================================================

companies = ["Walmart", "Amazon", "UnitedHealth\nGroup", "Apple", "CVS Health",
             "Berkshire\nHathaway", "Alphabet", "ExxonMobil", "McKesson", "Cencora"]
revenue = [681, 638, 400, 391, 373, 371, 350, 350, 309, 294]

fig, ax = plt.subplots(figsize=(9.5, 5))
colors = ["#2563eb"] * 10
colors[0] = "#1e3a8a"  # highlight the #1 company
bars = ax.barh(companies[::-1], revenue[::-1], color=colors[::-1], height=0.6)

for bar, val in zip(bars, revenue[::-1]):
    ax.text(bar.get_width() + 8, bar.get_y() + bar.get_height() / 2, f"${val}B",
            va="center", fontsize=10, fontweight="bold", color="#111827")

ax.set_xlabel("Revenue (US$ Billions)", fontsize=11)
ax.set_title("Top 10 U.S. Companies by Revenue", fontsize=15, fontweight="bold", pad=15)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.set_xlim(0, 760)
plt.tight_layout()
plt.savefig("chart1_top10_revenue.png", dpi=200)
plt.show()


# ==================================================================
# CHART 2 — Industry Distribution across Top 50 companies (Table 2)
# ==================================================================

from collections import Counter

industries = [
    "Communication Services", "Information Technology", "Information Technology", "Information Technology",
    "Consumer Discretionary", "Financials", "Communication Services", "Financials", "Energy", "Financials",
    "Health Care", "Communication Services", "Communication Services", "Consumer Staples", "Financials",
    "Financials", "Communication Services", "Information Technology", "Health Care", "Health Care",
    "Industrials", "Consumer Staples", "Health Care", "Financials", "Consumer Discretionary", "Financials",
    "Financials", "Health Care", "Energy", "Financials", "Consumer Staples", "Information Technology",
    "Communication Services", "Communication Services", "Financials", "Communication Services", "Financials",
    "Information Technology", "Health Care", "Industrials", "Energy", "Consumer Staples", "Consumer Staples",
    "Information Technology", "Consumer Discretionary", "Health Care", "Consumer Staples", "Industrials",
    "Information Technology", "Information Technology"
]
# NOTE: this list is just df['Industry'].tolist() from Table 2 (all 50 rows), typed out here
# so the script is self-contained. If you have the DataFrame, replace this whole list with:
#     industries = df['Industry'].tolist()

counts = Counter(industries)
items = sorted(counts.items(), key=lambda x: x[1], reverse=True)
labels = [i[0] for i in items]
vals = [i[1] for i in items]

fig, ax = plt.subplots(figsize=(9.5, 5))
bars = ax.bar(labels, vals, color="#2563eb", width=0.6)
bars[0].set_color("#1e3a8a")  # highlight the most common industry

for bar, v in zip(bars, vals):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.15, str(v),
            ha="center", fontsize=11, fontweight="bold", color="#111827")

ax.set_ylabel("Number of Companies", fontsize=11)
ax.set_title("Industry Distribution — Top 50 U.S. Companies\n(by Net Profit, scraped dataset)",
             fontsize=14, fontweight="bold", pad=15)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.xticks(rotation=30, ha="right", fontsize=9)
ax.set_ylim(0, max(vals) + 1.5)
plt.tight_layout()
plt.savefig("chart2_industry_dist.png", dpi=200)
plt.show()


# ==========================================================
# CHART 3 — Net Profit vs. Revenue scatter (Table 2, 50 rows)
# ==========================================================

data = [
    ("Alphabet", 124.3, 385.5), ("Apple", 112.0, 416.2), ("Microsoft", 104.9, 293.8),
    ("NVIDIA", 86.6, 165.2), ("Amazon", 76.5, 691.3), ("Berkshire Hathaway", 67.5, 372.1),
    ("Meta", 58.5, 189.5), ("JPMorgan Chase", 56.7, 277.7), ("Exxon Mobil", 30.0, 324.9),
    ("Bank of America", 28.3, 188.8), ("Johnson & Johnson", 25.1, 92.1), ("Comcast", 22.6, 123.3),
    ("AT&T", 22.2, 124.5), ("Walmart", 21.3, 693.2), ("Wells Fargo", 20.0, 122.6),
    ("Visa", 19.9, 40.0), ("Verizon", 19.8, 137.5), ("Broadcom", 18.9, 59.9),
    ("Eli Lilly", 18.4, 59.4), ("UnitedHealth Group", 17.6, 435.2), ("Uber", 16.6, 49.6),
    ("Procter & Gamble", 16.5, 84.9), ("Merck", 16.4, 63.6), ("Goldman Sachs", 15.8, 127.2),
    ("Home Depot", 14.6, 165.1), ("Mastercard", 14.3, 31.5), ("Morgan Stanley", 14.1, 113.0),
    ("Abbott Labs", 14.0, 43.8), ("Chevron", 13.7, 187.7), ("Citigroup", 13.4, 168.3),
    ("Coca-Cola", 13.0, 47.7), ("Oracle", 12.4, 59.0), ("T-Mobile US", 11.9, 85.8),
    ("Walt Disney", 11.6, 94.5), ("Progressive", 10.7, 85.2), ("Netflix", 10.4, 43.4),
    ("American Express", 10.4, 75.9), ("Cisco", 10.2, 56.7), ("Pfizer", 9.8, 62.8),
    ("Caterpillar", 9.3, 64.7), ("ConocoPhillips", 9.2, 57.8), ("Altria", 8.8, 23.4),
    ("Philip Morris Intl", 8.6, 40.0), ("Micron", 8.5, 37.4), ("McDonald's", 8.4, 26.3),
    ("Gilead Sciences", 8.1, 29.1), ("Costco", 8.1, 275.2), ("General Electric", 8.1, 43.9),
    ("IBM", 7.9, 65.4), ("MicroStrategy", 7.7, 0.5),
]
# NOTE: this is (Company Name, Net Profit, Revenue) for all 50 rows of Table 2.
# With the DataFrame, you'd instead do:
#     names, profit, revenue = df['Company Name'], df['Net Profit ($B)'].astype(float), df['Revenue ($B)'].astype(float)

profit = [d[1] for d in data]
revenue_vals = [d[2] for d in data]

fig, ax = plt.subplots(figsize=(9.5, 5.5))
ax.scatter(revenue_vals, profit, s=70, color="#2563eb", alpha=0.75,
           edgecolor="white", linewidth=0.6, zorder=3)

highlight = ["Amazon", "Walmart", "Apple", "Alphabet", "UnitedHealth Group", "MicroStrategy"]
for n, pr, rv in data:
    if n in highlight:
        ax.scatter([rv], [pr], s=90, color="#dc2626", edgecolor="white", linewidth=0.8, zorder=4)
        ax.annotate(n, (rv, pr), xytext=(6, 6), textcoords="offset points",
                    fontsize=8.5, fontweight="bold", color="#111827")

ax.set_xlabel("Revenue (US$ Billions)", fontsize=11)
ax.set_ylabel("Net Profit (US$ Billions)", fontsize=11)
ax.set_title("Net Profit vs. Revenue — Top 50 U.S. Companies", fontsize=14, fontweight="bold", pad=15)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.tight_layout()
plt.savefig("chart3_scatter.png", dpi=200)
plt.show()
```
